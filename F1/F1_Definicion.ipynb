# 1. Contextualización

Airbnb se ha consolidado como una de las principales plataformas de alojamiento temporal a nivel mundial. En Santiago existe una amplia oferta de propiedades con diferentes características, ubicaciones y niveles de actividad.

La diversidad de alojamientos disponibles genera oportunidades para potenciales nuevos anfitriones, pero también dificulta la identificación de las mejores alternativas para ingresar al mercado de manera competitiva.

Mediante el análisis de datos se busca transformar la información disponible en conocimiento útil para apoyar la toma de decisiones de futuros anfitriones.

# 2. Problemática

Los potenciales nuevos anfitriones que desean incorporarse al mercado Airbnb en Santiago enfrentan dificultades para identificar qué comunas, tipos de alojamiento y características presentan mejores oportunidades.

Aunque existe abundante información disponible, esta requiere ser recopilada, organizada y analizada para transformarse en evidencia que apoye decisiones fundamentadas.

# 3. Objetivos

## Objetivo General

Analizar el mercado Airbnb de Santiago para identificar oportunidades y características relevantes para potenciales nuevos anfitriones.

## Objetivos Específicos

1. Obtener el conjunto de datos seleccionado desde Airbnb.
2. Configurar un entorno reproducible para el proyecto.
3. Verificar la disponibilidad y estructura general de los datos.
4. Preparar la base técnica para las siguientes fases del proyecto.
5. Establecer un flujo de trabajo colaborativo mediante Git y GitHub.
